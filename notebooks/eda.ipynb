```python
from matplotlib.patches import Patch

features = df.drop(columns=['id', 'flag'], errors='ignore')
correlations = features.corrwith(df['flag'])
top_correlations = correlations.abs().sort_values(ascending=False)
res = correlations.loc[top_correlations.index].head(15)

plt.figure(figsize=(13, 8))

colors = {name: 'C0' if val > 0 else 'C1' for name, val in res.items()}

sns.barplot(y=res.index, x=abs(res.values), palette=colors)

legend_elements = [Patch(facecolor='C0', label='Положительная'),
                   Patch(facecolor='C1', label='Отрицательная')]
plt.legend(handles=legend_elements, loc='lower right')

plt.xlabel('Коэффициент корреляции')
plt.ylabel('Признак')
plt.title('Топ 15 признаков по корреляции с целевой переменной')
plt.axvline(x=0, color='black', linestyle='-', linewidth=0.5)
plt.show()
```
